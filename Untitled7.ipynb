```python
#4
while True:
    try:
        num = int(input("digite um numero: "))
        if num < 0:
            print("digite um numero positivo.")
        else:
            break
    except:
        print("digite um numero inteiro.")

fatorial = 1
if num == 0:
    print(f"o fatorial de {num} é 1")
elif num > 0:
    for i in range(1, num + 1):
        fatorial *= i
    print(f"o fatorial de {num} é {fatorial}")


```

```python
#5
# notas de:
#R$ 100
#R$ 50
#R$ 20
#R$ 10
#R$ 5
#R$ 2
#R$ 1

valor = int (input("Digite o valor que quer sacar: "))

valor_rest = valor

for nota in (100, 50, 20, 10, 5, 2, 1):

    quantidade_notas = valor_rest // nota

    if quantidade_notas > 0:
        print(f"{quantidade_notas} nota(s) de R$ {nota}")
    valor_rest %= nota
```
